### Implementation & Rationale


#### How did you implement this topic and why did you use it?

**[We treat ARIMAX as a strong baseline (test RMSE ≈ 0.2588), and use hyperparameter tuning to check whether better-tuned neural models can close this gap. For the LSTM, we run a grid search over hidden size, number of layers, dropout rate, learning rate, batch size, and max epochs, and rank all variants by validation RMSE.

We then perform a similar grid search for a stand-alone MLP (the post-LSTM refiner), tuning layer width/depth, dropout, and learning rate to see if an improved MLP design can fix the weak performance from our initial experiments. Finally, we jointly tune the full LSTM+MLP hybrid, searching over both LSTM and MLP hyperparameters to find the best multimodal combination of price, fundamentals, and news.

Across all searches, we use a fixed train/validation split, compare models using validation RMSE, and set random seeds for reproducibility. This systematic grid-search tuning goes beyond a single hand-picked architecture and provides a fair, apples-to-apples comparison between neural models and the ARIMAX baseline.]**

### Implementation & Rationale


#### How did you implement this topic and why did you use it?

**[we prefer Plotly for the time-series plots in this project because it gives an analyst experience that actually feels useful. It is closer to what you’d get on a trading terminal and that matters when you try to understand high-frequency stock behavior. In practice the interactive features make a big difference. You can inspect exact price+MA values at any minute without cluttering the figure. The smooth zooming and panning help when you want to visually validate patterns across months or years — static PNGs hide a lot of short bursts and volatility clustering that are obvious when you scrub the timeline.

Plotly also handles dense overlays (price + MA20 + MA60, multiple indicators) much better: lines remain sharp and tooltips keep the chart readable even with tens of thousands of points. It renders nicely inside notebooks and produces shareable HTMLs which we can hand to teammates or include in the project report. Yes, Matplotlib is fine for publication-style static figures, but for exploratory work — where we need to test hypotheses quickly and show interactive evidence to each other — Plotly is simply more efficient. So in short: it speeds up EDA, reduces friction when diagnosing model failures, and provides an analyst-friendly UX that matches the problem we’re solving.]**
